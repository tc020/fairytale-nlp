```python
#Zum ausführen die 3.13er venv nutzen, sonst gibts Probleme mit Stanza

import spacy
import stanza
from flair.models import SequenceTagger
from flair.data import Sentence
from collections import Counter

nlp_spacy = spacy.load("de_core_news_lg")

stanza.download("de")
nlp_stanza = stanza.Pipeline("de", processors="tokenize,pos", verbose=False)

flair_tagger = SequenceTagger.load("de-pos")

STTS_TO_UPOS = {
    "NN": "NOUN", "NE": "PROPN",
    "VVFIN": "VERB", "VVINF": "VERB", "VVPP": "VERB", "VVIMP": "VERB",
    "VAFIN": "AUX",  "VAINF": "AUX",  "VAPP": "AUX",
    "VMFIN": "AUX",  "VMINF": "AUX",  "VMPP": "AUX",
    "ADJA": "ADJ",   "ADJD": "ADJ",
    "ART": "DET",    "PDAT": "DET",   "PIAT": "DET",  "PPOSAT": "DET",
    "PPER": "PRON",  "PRF": "PRON",   "PPOSS": "PRON","PRELS": "PRON",
    "PWS": "PRON",   "PWAV": "ADV",   "PWAT": "DET",
    "ADV": "ADV",    "PROAV": "ADV",
    "APPR": "ADP",   "APPRART": "ADP","APPO": "ADP",  "APZR": "ADP",
    "KON": "CCONJ",  "KOUS": "SCONJ", "KOUI": "SCONJ","KOKOM": "CCONJ",
    "CARD": "NUM",
    "ITJ": "INTJ",
    "PTKNEG": "PART","PTKZU": "PART", "PTKVZ": "PART","PTKANT": "PART",
    "XY": "X",       "FM": "X",
    "$.": "PUNCT",   "$,": "PUNCT",   "$(": "PUNCT",
    "TRUNC": "X",
}

def get_spacy(text):
    doc = nlp_spacy(text)
    return [(t.text, t.pos_) for t in doc]


def get_stanza(text):
    doc = nlp_stanza(text)
    out = []
    for s in doc.sentences:
        for w in s.words:
            out.append((w.text, w.upos))
    return out


def get_flair(text):
    sent = Sentence(text)
    flair_tagger.predict(sent)

    out = []
    for t in sent:
        tag = t.get_label("pos").value
        upos = STTS_TO_UPOS.get(tag, tag)
        out.append((t.text, upos))
    return out

def majority_vote(spacy_pos, stanza_pos, flair_pos):

    votes = [spacy_pos, stanza_pos, flair_pos]
    count = Counter(votes)

    #Mehrheitsentscheidung
    most_common = count.most_common()

    #klare Mehrheit
    if len(most_common) == 1:
        return most_common[0][0]

    if most_common[0][1] > 1:
        return most_common[0][0]

    #Tie => spaCy gewinnt
    return spacy_pos

def ensemble_conll(text):

    spacy_tokens = get_spacy(text)
    stanza_tokens = get_stanza(text)
    flair_tokens = get_flair(text)

    #Alignment (einfach positionsbasiert)
    max_len = min(len(spacy_tokens), len(stanza_tokens), len(flair_tokens))

    conll = []

    for i in range(max_len):

        token = spacy_tokens[i][0]

        spacy_pos = spacy_tokens[i][1]
        stanza_pos = stanza_tokens[i][1]
        flair_pos = flair_tokens[i][1]

        final_pos = majority_vote(spacy_pos, stanza_pos, flair_pos)

        #CoNLL-BIO Format:
        #Token POS BIO (NER placeholder)
        conll.append(f"{token}\t{final_pos}\tO")

    return "\n".join(conll)

#text = "Der schnelle braune Fuchs springt über den faulen Hund."

def ensemble_conll_sents(text):
    # Sätze mit spaCy erkennen
    doc = nlp_spacy(text)
    sents = [sent.text.strip() for sent in doc.sents]

    blocks = []
    for sent in sents:
        block = ensemble_conll(sent)  #bestehendes Ensemble
        blocks.append(block)

    #Sätze mit Leerzeile trennen
    return "\n\n".join(blocks)

result = ensemble_conll_sents(text)

with open("train-andersen.conll", "w", encoding="utf-8") as f: #<<< Output-Name anpassen!
    f.write(result)
```

## Finetuning NER

```python
"""
spaCy Finetuning-Pipeline für NER + POS
Basiert auf dem vortrainierten deutschen Modell de_core_news_lg.
Format der CoNLL-Datei: Token | POS-Tag | NER-Tag (BIO-Schema)
Leerzeilen = Satzgrenzen
"""

import spacy
from spacy.tokens import DocBin, Doc
import random
from pathlib import Path


# ─────────────────────────────────────────────
# 1. CoNLL-Datei einlesen (unverändert)
# ─────────────────────────────────────────────

def read_conll(filepath: str):
    """
    Liest eine 3-spaltige CoNLL-Datei ein.
    Gibt eine Liste von Sätzen zurück, jeder Satz ist eine Liste von
    (token, pos, ner)-Tupeln.
    """
    sentences = []
    current = []

    with open(filepath, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if line.strip() == "":
                if current:
                    sentences.append(current)
                    current = []
            else:
                parts = line.split("\t")
                if len(parts) == 3:
                    token, pos, ner = parts
                    current.append((token, pos, ner))

    if current:
        sentences.append(current)

    return sentences


# ─────────────────────────────────────────────
# 2. BIO-Tags → spaCy Entity-Spans (unverändert)
# ─────────────────────────────────────────────

def bio_to_spans(tokens, ner_tags):
    spans = []
    start = None
    current_label = None
    char_offset = 0
    offsets = []

    for token in tokens:
        offsets.append(char_offset)
        char_offset += len(token) + 1

    for i, tag in enumerate(ner_tags):
        if tag.startswith("B-"):
            if start is not None:
                spans.append((start, offsets[i] - 1, current_label))
            start = offsets[i]
            current_label = tag[2:]
        elif tag.startswith("I-") and start is not None:
            pass
        else:
            if start is not None:
                end = offsets[i] - 1 if i > 0 else char_offset
                spans.append((start, end, current_label))
                start = None
                current_label = None

    if start is not None:
        spans.append((start, char_offset - 1, current_label))

    return spans


# ─────────────────────────────────────────────
# 3. spaCy DocBin erstellen
# ÄNDERUNG: nlp ist jetzt das vortrainierte Modell,
#           wir nutzen dessen Vocab für korrekte Tokenisierung
# ─────────────────────────────────────────────

def sentences_to_docbin(sentences, nlp):
    doc_bin = DocBin(store_user_data=True)

    for sent in sentences:
        sent = [t for t in sent if t[0].strip() != ""]   #leere/Whitespace-Tokens überspringen
        if not sent:
            continue
        tokens  = [t[0] for t in sent]
        pos_tags = [t[1] for t in sent]
        ner_tags = [t[2] for t in sent]

        spaces = [True] * len(tokens)
        spaces[-1] = False

        #Vocab des vortrainierten Modells verwenden
        doc = Doc(nlp.vocab, words=tokens, spaces=spaces)

        #POS-Tags setzen
        for token, pos in zip(doc, pos_tags):
            token.pos_ = pos
            token.tag_ = pos

        #NER-Spans setzen
        spans = bio_to_spans(tokens, ner_tags)
        ents = []
        for start_char, end_char, label in spans:
            span = doc.char_span(start_char, end_char, label=label)
            if span is not None:
                ents.append(span)
        doc.ents = ents

        doc_bin.add(doc)

    return doc_bin


# ─────────────────────────────────────────────
# 4. config.cfg für Finetuning
# ÄNDERUNGEN:
#   - base_model gesetzt → lädt vortrainierte Gewichte
#   - morphologizer entfernt → war nutzlos (MORPH_ACC=0)
#   - patience/max_steps angepasst für kleine Datenmenge
#   - NER wird neu initialisiert (eigene Labels)
#   - tok2vec wird vom Basismodell übernommen
# ─────────────────────────────────────────────

CONFIG = """
[paths]
train = "corpus/train.spacy"
dev = "corpus/dev.spacy"
vectors = null
init_tok2vec = null

[system]
gpu_allocator = null
seed = 42

[nlp]
lang = "de"
pipeline = ["tok2vec", "tagger", "ner"]
batch_size = 1000

[components]

[components.tok2vec]
source = "de_core_news_lg"

[components.tagger]
factory = "tagger"

[components.tagger.model]
@architectures = "spacy.Tagger.v2"
nO = null

[components.tagger.model.tok2vec]
@architectures = "spacy.Tok2VecListener.v1"
width = 96
upstream = "tok2vec"

[components.ner]
factory = "ner"

[components.ner.model]
@architectures = "spacy.TransitionBasedParser.v2"
state_type = "ner"
extra_state_tokens = false
hidden_width = 64
maxout_pieces = 2
use_upper = true
nO = null

[components.ner.model.tok2vec]
@architectures = "spacy.Tok2VecListener.v1"
width = 96
upstream = "tok2vec"

[training]
train_corpus = "corpora.train"
dev_corpus = "corpora.dev"
seed = ${system.seed}
gpu_allocator = ${system.gpu_allocator}
dropout = 0.1
patience = 0
max_steps = 3000
eval_frequency = 600

[training.batcher]
@batchers = "spacy.batch_by_words.v1"
discard_oversize = false
tolerance = 0.2

[training.batcher.size]
@schedules = "compounding.v1"
start = 100
stop = 1000
compound = 1.001

[training.logger]
@loggers = "spacy.ConsoleLogger.v1"
progress_bar = true

[training.optimizer]
@optimizers = "Adam.v1"
beta1 = 0.9
beta2 = 0.999
L2_is_weight_decay = true
L2 = 0.01
grad_clip = 1.0
use_averages = false
eps = 0.00000001
learn_rate = 0.0005

[training.score_weights]
ents_f = 0.5
pos_acc = 0.0
tag_acc = 0.5

[corpora]

[corpora.train]
@readers = "spacy.Corpus.v1"
path = ${paths.train}
max_length = 0

[corpora.dev]
@readers = "spacy.Corpus.v1"
path = ${paths.dev}
max_length = 0

[initialize]
vectors = "de_core_news_lg"
init_tok2vec = ${paths.init_tok2vec}
vocab_data = null
"""

#vectors = ${paths.vectors}

# ─────────────────────────────────────────────
# 5. Haupt-Pipeline
# ÄNDERUNG: spacy.load() statt spacy.blank()
# ─────────────────────────────────────────────

def main(conll_path: str = "train-aschenputtel.conll"):
    dateien = [
        "train-aschenputtel.conll",
        "train-dornroeschen.conll",
        "train-frau_holle.conll",
        "train-froschkoenig.conll",
        "train-hansel_und_gretel.conll",
        "train-rapunzel.conll",
        "train-rotkäppchen.conll",
        "train-rumpelstilzchen.conll", 
        "train-schneewittchen.conll",
        "train-grimm1.conll", 
        "train-grimm2.conll", 
        "train-arabian.conll",
        "train-japanese.conll", 
        "train-andersen.conll", 
    ]

    print("Lese CoNLL-Dateien ein ...")
    alle_saetze = []
    for datei in dateien:
        saetze = read_conll(datei)
        alle_saetze.extend(saetze)
        print(f"   => Gefunden: {datei} mit {len(saetze)} Sätzen")

    print(f"   => Gesamt: {len(alle_saetze)} Sätze geladen")

    """
    #Block für eine Datei
    print("📖 Lese CoNLL-Datei ein ...")
    sentences = read_conll(conll_path)
    print(f"   => {len(sentences)} Sätze geladen")
    """

    random.seed(42)
    random.shuffle(alle_saetze)
    split = int(len(alle_saetze) * 0.8)
    train_sents = alle_saetze[:split]
    dev_sents   = alle_saetze[split:]
    print(f"   => Train: {len(train_sents)} | Dev: {len(dev_sents)}")

    # ── ÄNDERUNG: vortrainiertes Modell laden ──
    print("Lade vortrainiertes Modell de_core_news_lg ...")
    nlp = spacy.load("de_core_news_lg")

    corpus_dir = Path("corpus")
    corpus_dir.mkdir(exist_ok=True)

    print("Erstelle .spacy Dateien ...")
    train_db = sentences_to_docbin(train_sents, nlp)
    dev_db   = sentences_to_docbin(dev_sents, nlp)
    nlp.vocab.to_disk(corpus_dir / "vocab")  #Vocab explizit mitgeben
    train_db.to_disk(corpus_dir / "train.spacy")
    dev_db.to_disk(corpus_dir / "dev.spacy")
    print("   => corpus/train.spacy und corpus/dev.spacy gespeichert")

    config_path = Path("config.cfg")
    config_path.write_text(CONFIG.strip(), encoding="utf-8")
    print("config.cfg für Finetuning gespeichert")

    print("\nVorbereitung abgeschlossen!")
    print("\n── Schritt 1: Modell installieren (falls noch nicht geschehen) ──")
    print("   python -m spacy download de_core_news_sm")
    print("\n── Schritt 2: Training starten ──")
    print("   python -m spacy train config.cfg --output output/")
    print("\n── Schritt 3: Modell testen ──")
    print("   python test_model.py")
    print("\nBestes Modell liegt in: output/model-best/")


if __name__ == "__main__":
    main(conll_path="train-aschenputtel.conll")
```
